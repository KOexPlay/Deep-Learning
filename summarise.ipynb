```python
paragraph = '''Karate (空手) Japanese pronunciation: [kaɾate]; Okinawan pronunciation: [kaɾati]), also karate-do (空手道, Karate-dō), is a martial art developed in the Ryukyu Kingdom. 
It developed from the indigenous Ryukyuan martial arts (called te (手), "hand"; tī in Okinawan) under the influence of Chinese martial arts.
While modern karate is primarily a striking art that uses punches and kicks, traditional karate training also employs throwing and joint locking techniques.
A karate practitioner is called a karate-ka (空手家). Beginning in the 1300s, early Chinese martial artists brought their techniques to Okinawa.
Despite the Ryukyu Kingdom being turned into a puppet state by Japanese samurai in 1609, after the Invasion of Ryukyu, its cultural ties to China remained strong.
Since Ryukyuans were banned from carrying swords under samurai rule, groups of young aristocrats created unarmed combat methods as a form of resistance, combining Chinese and local styles of martial arts.
Training emphasized self-discipline. This blend of martial arts became known as kara-te 唐手, which translates to "Chinese hand."
Initially, there were no uniforms, colored belts, ranking systems, or standardized styles.
Many elements essential to modern karate were actually incorporated a century ago.'''
print(paragraph)
```

```python
import nltk
from nltk.tokenize import sent_tokenize
```

```python
sentences = sent_tokenize(paragraph)
print(sentences)
```

```python
import re
newSents = []

for sent in sentences:
    sent = re.sub(r"[^\w\s]", "", sent.lower())
    newSents.append(sent)
print(newSents)
```

```python
from nltk.corpus import stopwords

fillerWords = stopwords.words("english")
processedSents = []

for sent in newSents:
    words = sent.split(" ")
    processedSents.append(words)

print(processedSents)
```

```python
newProcessedSents = []

for s in processedSents:
    #print(s)
    newSent = []
    for w in s:
        if w not in fillerWords:
            newSent.append(w)
    newProcessedSents.append(newSent)
print(newProcessedSents)
```

```python
from gensim.models import Word2Vec

embedding = Word2Vec(newProcessedSents, vector_size = 1, min_count = 1, epochs = 100)
print(embedding)
```

```python
embeddingList = []

for s in newProcessedSents:
    vectors = []
    for w in s:
        vectors.append(embedding.wv[w][0])
    embeddingList.append(vectors)
print(embeddingList)
```

```python
lengths = []
for s in embeddingList:
    lengths.append(len(s))
    #print(len(s))
maxLength = max(lengths)
```

```python
import numpy as np
newEmbeddingList = []

for s in embeddingList:
    newEmbeddingList.append(np.pad(s, (0, maxLength-len(s))))
print(newEmbeddingList)

```

```python
similarity_matrix = np.zeros((len(newEmbeddingList), len(newEmbeddingList)))
print(similarity_matrix)
```

```python
from scipy import spatial


for s in range(len(newEmbeddingList)):
    for w in range(len(newEmbeddingList)):
        similarity_matrix[s][w] = (1-spatial.distance.cosine(newEmbeddingList[s], newEmbeddingList[w]))
print(similarity_matrix)
```

```python
import networkx

graph = networkx.from_numpy_array(similarity_matrix)
networkx.pagerank(graph, tol = 1000.0)
```
