```python
import cv2
import sys
import numpy as np
```

```python
coco_names = r"C:\Users\kaige\python projects\Deep Learning\coco.names"
configYOLO4 = r"C:\Users\kaige\python projects\Deep Learning\yolov4.cfg"
weightsYOLO4 = r"C:\Users\kaige\python projects\Deep Learning\yolov4.weights"
```

```python
file = open(coco_names, "r")
cocoList = []
tempList = file.readlines()
for item in tempList:
    cocoList.append(item.strip("\n"))
print(cocoList)
```

```python
YOLO = cv2.dnn.readNet(weightsYOLO4, configYOLO4)#weight, config
```

```python
layerNames = YOLO.getLayerNames()
print(layerNames)
```

```python
outputLayers = YOLO.getUnconnectedOutLayersNames()
print(outputLayers)
```

```python
webcam = cv2.VideoCapture(0)
while True:
    (_, im) = webcam.read()
    height, width, channels = im.shape
    imageFormat = cv2.dnn.blobFromImage(im, 0.5, (height, width), (0, 0, 0), True)
    YOLO.setInput(imageFormat)
    output = YOLO.forward(outputLayers)

    # Detection
    labels = []
    bounding_boxes = []
    accuracy = []

    for i in output:
        
        for detection in i:
            values = detection[5:]
            maxIndex = np.argmax(values)
            maxValue = values[maxIndex]
            if maxValue >= 0.5:
                x_cord = int(detection[0] * width)
                y_cord = int(detection[1] * height)
                width2 = int(detection[2] * width)
                height2 = int(detection[3] * height)
                bounding_boxes.append([x_cord, y_cord, width2, height2])
                labels.append(maxIndex)
                accuracy.append(maxValue)
    cv2.dnn.NMSBoxes(bounding_boxes, accuracy, 0.5, 0.4)
    for i in range(len(bounding_boxes)):
        bx, by, bwidth, bheight = bounding_boxes[i]
        name = cocoList[labels[i]]
        cv2.putText(im, name, (bx, by),cv2.FONT_ITALIC, 0.5, (20, 255, 0), 1)
        cv2.rectangle(im, (bx,  by), (bx +bwidth, by + bheight), (0, 255, 0), 1)
    cv2.imshow("display", im)
    key = cv2.waitKey(1)
    if key == 27:
        break
webcam.release()
cv2.destroyAllWindows()
```
